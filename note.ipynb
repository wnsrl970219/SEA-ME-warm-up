# My note for confusing c++ concept

## 1. Never return reference to locally variables
when you want a function take inputs and some manipualted input..
  * local variable not exists when function exits :thumbsdown:
  
  ```cpp
  #include <iostream >
  using namespace std;  
  int& MultiplyBy10 (int num) { // retval is created
  int retval = 0;
  retval = 10 * num;
  return retval;
  } // retval is destroyed , it's not accesisble anymore  
   int main () {
   int out = MultiplyBy10 (10);
   cout << "out is " << out << endl;
   return 0;
   }
  ```

  * using static local variable as return :thumbsup:❓-avoid if possible
  
    ```cpp
    int& function(int arg) {
    static int result = arg * 2;  // Static variable persists across function calls
    return result;
    }
    ```
  * using input argument reference as return :thumbsup::thumbsup:
  
    ```cpp
    int& function(int& arg) {
    arg *= 2;  // Modify the argument directly
    return arg;  // Return a reference to the modified argument
    }
    ```

## 3. Function overloading
Can not overload based on return type

  * overload with argument type :thumbsup:
  
    ```cpp
    #include <iostream >
    #include <string >
    using namespace std;

    string TypeOf(int) { return "int"; }
    
    string TypeOf(const string &) { return "string";}

    int add(int x, int y);          
    double add(double x, double y); 

    int add(int x, int y, int z)
    {
        return x + y + z;
    }

    int main () {
    cout << TypeOf (1) << endl;
    cout << TypeOf("hello") << endl;
    return 0;
      }
    ```
  * overload with argument type :thumbsdown:
    ```cpp
    int getRandomValue();
    double getRandomValue();
    ```

## 4. Namepace
Avoid using namespace \<name\>

  * there is already name std(standard library) :thumbsdown:
  
    ```cpp
    # include <cmath >
    # include <iostream >
    using namespace std; // std namespace is used
    // Self -defined function power shadows std::pow
    double pow(double x, int exp) {
    double res = 1.0;
    for (int i = 0; i < exp; i++) {
    res *= x;
    }
    return res;
    }

    int main () {
    cout << "2.0^2 = " << pow (2.0 , 2) << endl;
    return 0;
    }
    ```
  compile error

  * write it explicitylt :thumbsup:
  
    ```cpp
    # include <cmath >
    # include <iostream >
    using std :: cout; // Explicitly use cout.
    using std :: endl; // Explicitly use endl.
    // Self -defined function power shadows std::pow
    double pow(double x, int exp) {
    double res = 1.0;
    for (int i = 0; i < exp; i++) {
    res *= x;
    }
    return res;
    }

    int main () {
    cout << "2.0^2 = " << pow (2.0 , 2) << endl;
    return 0;
    }
    ```
  compile error
